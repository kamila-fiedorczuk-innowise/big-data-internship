**1.** Given a random array, change the sign of elements whose values ​​are between 3 and 8

```python
a = np.random.randint(0, 15, 20)
print(a)
mask = (a > 3) & (a < 8)
a[mask] *= -1
print(a)
```

**4.** Given two arrays, A (8x3) and B (2x2), find rows in A that contain elements from each row in B, regardless of the order of the elements in B.

```python
A = np.random.randint(0, 5, (8, 3))
print(A)
B = np.random.randint(0, 5, (2, 2))
print(B)

m0 = np.isin(A, B[0]).any(axis=1)
m1 = np.isin(A, B[1]).any(axis=1)
mask = m0 & m1
print(mask)
print(A[mask])
```

* __Task 1__: Calculate the product of nonzero elements on the diagonal of a rectangular matrix.  
 For example, for X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) answer is 3.

```python
# Python
L = [[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]
result = 1
for i in range(min(len(L), len(L[0]))):
    if L[i][i] != 0:
        result *= L[i][i]
print(result)
```

```python
# NumPy
N = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]])
d = np.diag(N)
print(np.prod(d[d != 0]))
```

* __Task 3__: Find the maximum element in the vector x among the elements preceded by zero. 
 For example, for x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0]) answer is 5.

```python
# Python
x = [6, 2, 0, 3, 0, 0, 5, 7, 0]
score = []
for i in range(1, len(x)):
    if x[i - 1] == 0:
        score.append(x[i])
print(max(score))
```

```python
# NumPy
x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0])
print(x[1:][x[:-1] == 0].max())
```

* __Task 4__: Implement run-length encoding. For some vector x, it is necessary to return a tuple of two vectors of the same length. The first one contains numbers and the second one contains how many times they need to be repeated.  
 For example, for x = np.array([2, 2, 2, 3, 3, 3, 5]) answer is (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
# Python
x = [2, 2, 2, 3, 3, 3, 5]
numbers = []
count = []
for i in range(len(x)):
    if numbers and x[i] == numbers[-1]:
        count[-1] += 1
    else:
        numbers.append(x[i])
        count.append(1)
print((numbers, count))
```

```python
# NumPy
x = np.array([2, 2, 2, 3, 3, 3, 5])

is_change = x[1:] != x[:-1]
change_idx = np.flatnonzero(is_change) + 1
run_starts = np.concatenate(([0], change_idx))
values = x[run_starts]
run_lengths = np.diff(np.concatenate((run_starts, [len(x)])))

print((values, run_lengths))
```

* __Task 5__: Two samples of objects are given - X and Y. Calculate the matrix of Euclidean distances between objects. Compare with function scipy.spatial.distance.cdist by operating speed.

```python
import math

from scipy.spatial.distance import cdist

X = np.random.rand(200, 3)
Y = np.random.rand(300, 3)
points_a_list = X.tolist()
points_b_list = Y.tolist()
```

```python
# Python
def dist_python(points_a, points_b):
    distances = []
    for pa in points_a:
        row = []
        for pb in points_b:
            squared_sum = 0
            for coord_a, coord_b in zip(pa, pb):
                squared_sum += (coord_a - coord_b) ** 2
            row.append(math.sqrt(squared_sum))
        distances.append(row)
    return distances


distances_py = dist_python(points_a_list, points_b_list)
print(np.allclose(np.array(distances_py), cdist(X, Y)))
%timeit dist_python(points_a_list, points_b_list)
```

```python
# NumPy
def dist_numpy(points_a, points_b):
    return np.sqrt(
        ((points_a[:, np.newaxis, :] - points_b[np.newaxis, :, :]) ** 2).sum(axis=2)
    )


Dis_np = dist_numpy(X, Y)
print(Dis_np.shape)
print(np.allclose(Dis_np, cdist(X, Y)))
%timeit dist_numpy(X, Y)
%timeit cdist(X, Y)
```

*   The pure Python implementation takes ~25 ms
*   the vectorized NumPy version ~1.4 ms (about 18× faster)
*   scipy’s cdist ~0.09 ms (about 15× faster than NumPy and ~275× faster than pure Python)

All three give the same result.
